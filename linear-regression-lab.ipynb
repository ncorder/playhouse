```r
ybar <- sum(y) / length(y)
ybar
```

```r
x - xbar
```

```r
(x - xbar)^2
```

```r
Sxx <- sum((x-xbar)^2)
Sxx
```

```r
Sxy <- sum((x-xbar)*(y-ybar))
Sxy
```

```r
# calculating the slope
beta1 <- Sxy / Sxx
beta1
```

```r
# calculating the intercept
beta0 <- ybar - beta1*xbar
beta0
```
